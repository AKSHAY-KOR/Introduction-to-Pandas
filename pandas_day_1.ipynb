```python
ser1 =  pd.Series([1,2,3,4,5])
ser2 = pd.Series([1,2,3,4,5],index=[2,3,4,5,6])
print(ser1)
print(ser2)
```

```python
#get the values of the 2 from the ser1 and ser2 
ser1[1]
```

```python
#ser2 at 3rd index we have the value of 2
ser2[3]#by default it consider the default explicit indexes
```

```python
#get the values of 3,4,5 using slicing
ser1[2:5]
```

```python
#at ser2 the values of 1,4,5 has at indexes 2,3,4 so we use 2:5 by the slicing
print(ser2[2:5])#here explicit indexing not taking default implicit indexing taking as default for slicing
print()
print(ser2[4:7])#its not taking explicit indexing
```

```python
#using advanced indexing getthe values of the 1,4,5
ser1[[0,3,4]]
```

```python
ser2[[2,5,6]]#pandas considering the explicit as default but in slicing it choosing implicit
```

```python
#incase of the advanced indexing and indexing pandas choosing the explicit indexing as default
#where as slicing pandas choosing the implicit index as the default
# advanced indexing-->explicit
# indexing-->explicit
# slicing-->implicit
```

```python
#to get rid of the confusing pandas does not depend on the default index and slicing in pandas we use 
#.loc() for the explicit
#.iloc() #for the implicit
```
